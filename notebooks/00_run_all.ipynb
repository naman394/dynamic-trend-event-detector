# Notebook 00 — Master Run: Full Pipeline

Execute this notebook to run the entire pipeline in correct order.

## Execution Order & Data Flow

| # | Notebook | Role | Depends on |
|---|---|---|---|
| 01 | EDA | Data inspection | raw data |
| 02 | Baseline TF-IDF | Keyword ranking + burst | 01 |
| 03 | Advanced ML LDA | Topic modelling + C_V | 01 |
| 08 | Deep Learning | SBERT K-Means + Semantic Velocity | 01 |
| 04 | GDELT Processor | Parse GKG snapshot | raw GDELT |
| 05 | GDELT Analysis | Theme/tone summary | 04 |
| **09** | **GDELT Verification + Impact** | **Verify ruptures, SBERT S_I scoring** | **08 + 05** |
| 07 | Visualise Results | Final dashboard | all above |

> **Key ordering rule:** Notebook 09 must run AFTER notebook 08 (it reads
> `reports/deep_learning/semantic_velocity.csv`) and after notebook 05
> (it reads `data/gdelt_processed.csv`).


```python
import os, subprocess, sys

if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')

NOTEBOOKS_DIR = 'notebooks'
NOTEBOOKS = [
    '01_eda.ipynb',
    '02_baseline_tfidf.ipynb',
    '03_advanced_ml_lda.ipynb',
    '04_gdelt_processor.ipynb',
    '05_gdelt_analysis.ipynb',
    '06_event_impact_scoring.ipynb',
    '07_visualize_results.ipynb',
    '08_deep_learning_kmeans.ipynb',
    '04_gdelt_processor.ipynb',
    '05_gdelt_analysis.ipynb',
    '09_gdelt_verification_impact.ipynb',  # AFTER 08 + 05
]
print('Will execute:', NOTEBOOKS)
```

```python
for nb in NOTEBOOKS:
    nb_path = os.path.join(NOTEBOOKS_DIR, nb)
    print(f'\n{"─"*60}')
    print(f'  Executing: {nb}')
    print('─' * 60)
    result = subprocess.run(
        [sys.executable, '-m', 'nbconvert', '--to', 'notebook',
         '--execute', '--inplace',
         '--ExecutePreprocessor.timeout=600',
         '--ExecutePreprocessor.kernel_name=python3',
         nb_path],
        capture_output=True, text=True
    )
    if result.returncode == 0:
        print(f'  ✅ Done')
    else:
        print(f'  ❌ Error:')
        print(result.stderr[-1000:] if result.stderr else '(no stderr)')

print('\n' + '='*60)
print('  Phase-1 pipeline complete.')
print('='*60)
```
